```python
import pandas as pd

# Load the dataset
df = pd.read_csv("/Bank Loan Data.csv")

# Create percentile groups for the variables we analyzed
df['CreditScoreGroup'] = pd.cut(
    df['CreditScore'],
    bins=[299, 399, 499, 599, 699, 799, 899],
    labels=['300-399', '400-499', '500-599', '600-699', '700-799', '800-899']
)

df['IncomeGroup'] = pd.qcut(
    df['Income'],
    q=4,
    labels=['Low', 'Lower-Middle', 'Upper-Middle', 'High']
)

df['LoanAmountGroup'] = pd.qcut(
    df['LoanAmount'],
    q=4,
    labels=['Low', 'Lower-Middle', 'Upper-Middle', 'High']
)

df['DTIGroup'] = pd.qcut(
    df['DTIRatio'],
    q=4,
    labels=['Lowest 25%', '25-50%', '50-75%', 'Highest 25%']
)

df['InterestRateGroup'] = pd.qcut(
    df['InterestRate'],
    q=4,
    labels=['Lowest 25%', '25-50%', '50-75%', 'Highest 25%']
)

df['AgeGroup'] = pd.qcut(
    df['Age'],
    q=4,
    labels=['Lowest 25%', '25-50%', '50-75%', 'Highest 25%']
)

df['LoanTermGroup'] = pd.qcut(
    df['LoanTerm'],
    q=4,
    labels=['Lowest 25%', '25-50%', '50-75%', 'Highest 25%']
)

print("All analysis groups recreated successfully.")
```

```python
remaining_defaults = df[
    ~(
        (df['IncomeGroup'] == 'Low') &
        (df['LoanAmountGroup'] == 'High') &
        (df['InterestRateGroup'] == 'Highest 25%')
    ) &
    (df['Default'] == 1)
]

print("Remaining defaults:", remaining_defaults.shape[0])
```

```python
# ============================================
# ANALYSIS OF THE REMAINING 95.08% OF DEFAULTS
# ============================================

analysis_columns = [
    'AgeGroup',
    'IncomeGroup',
    'LoanAmountGroup',
    'InterestRateGroup',
    'DTIGroup',
    'EmploymentType',
    'Education',
    'MaritalStatus',
    'LoanPurpose',
    'HasMortgage',
    'HasDependents',
    'HasCoSigner'
]

for column in analysis_columns:
    print("\n" + "=" * 60)
    print(column)
    print("=" * 60)

    counts = remaining_defaults[column].value_counts()
    percentages = remaining_defaults[column].value_counts(normalize=True) * 100

    result = pd.DataFrame({
        'DefaultCount': counts,
        'DefaultShare_%': percentages
    })

    print(result)
```
